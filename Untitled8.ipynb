```python
#import the necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

import joblib

print("libraries loaded sucessfully")
```

```python
#create a simple dataset
data = {
    'size_sq_ft': [500, 1000, 1500, 2000, 2500, 3000, 3500, 4000],
    'price_N': [200000, 350000, 500000, 650000, 800000, 950000, 1100000, 1250000]
}
df = pd.DataFrame(data)
df
```

```python
#visualize the data
plt.scatter(df['size_sq_ft'], df['price_N'])
plt.title('House prive vs size')
plt.xlabel('size(sqft)')
plt.ylabel('price(N)')
plt.show()
```

Preparing data for machine learning

```python
#split data into features (x) and target (y)
x =df[['size_sq_ft']]
y =df[['price_N']]

#split our data into training and testing
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size =0.2, random_state =42)
```

Building the model

```python
#create and train the model
model = LinearRegression()
model.fit(x_train,y_train)
```

Prediction

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

Understanding the model

```python
print(f'slope (a): {model.coef_[0]}')
print(f'intercept (b): {model.intercept_}')
```

The model's equation is : price = (slope"size) + intercept, Y = mx + b

```python
# Evaluate the model
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f'Mean squared error: {mse:.2f}')
print(f'Root mean squared error: {rmse:.2f}')
print(f'R-squared score: {r2:.2f}')
```

Visualize the model

```python
#visualize the predicted price vs the actual price
plt.scatter(x_test, y_test, color='red', label='Actual')
plt.plot (x_test, y_pred, color='blue', label='predicted')
plt.legend()
plt.show()
```

Saving the model

```python
#Saving the model
joblib.dump(model, 'House_price_prediction.joblib')
```

Import and use the model

```python
loaded_model = joblib.load('House_price_prediction.joblib')
```

```python
new_house = [[1800]]
predicted_price = loaded_model.predict(new_house)
print(f'predicted price: N {predicted_price[0]}')
```
